Create a line plot with multiple lines using Matplotlib.

```python
import matplotlib.pyplot as plt

# X-axis values
x = [1, 2, 3, 4, 5]

# Y-axis values for multiple lines
y1 = [1, 4, 9, 16, 25]
y2 = [2, 3, 5, 7, 11]
y3 = [5, 10, 15, 20, 25]

# Create the line plot
plt.plot(x, y1, label='y = x^2', color='blue')
plt.plot(x, y2, label='Prime numbers', color='red')
plt.plot(x, y3, label='Line 3')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Line Plot')

# Add a legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```

Create a histogram using Seaborn to visualize a distribution.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Generate a some random data (e.g., exam scores)
data = np.random.normal(loc=70, scale=10, size=100)

# Create a histogram using Seaborn
sns.histplot(data, bins=20, kde=True, color='skyblue')

# Add labels and title
plt.xlabel('Exam Scores')
plt.ylabel('Frequency')
plt.title('Score Distribution')

# Show the plot
plt.show()
```

Create a 3D scatter plot using Plotly.

```python
import plotly.express as px
import pandas as pd

# Create sample data
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [10, 15, 13, 17, 19],
    'z': [5, 6, 7, 8, 9]
})

# Create 3D scatter plot
fig = px.scatter_3d(df, x='x', y='y', z='z', title='3D Scatter Plot')

# Show the plot
fig.show()
```

Create a scatter plot using Matplotlib.

```python
import matplotlib.pyplot as plt

# Sample data
x = [10, 20, 30, 40, 50]
y = [5, 15, 25, 35, 45]

# Create a scatter plot
plt.scatter(x, y, color='b', marker='o')

# Add labels and title
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Sample Scatter Plot')

# Show the plot
plt.grid(True)
plt.show()
```
